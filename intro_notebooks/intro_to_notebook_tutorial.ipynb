## Why notebooks exist

Scientists rarely write a long program in one go. Instead we usually will perform the following:

1. write a code as a proof of concept or to try out new ideas,
2. run it,
3. look at the result,
4. change something,
5. repeat.

A notebook is a tool built for exactly this back-and-forth: small pieces of code, results shown
immediately, and room for notes explaining what you were thinking. It allows for both experiments *and* observations to live side by side.

It also remove the complexity related to Python development by simplifying environment setup, dependency management, and interactive experimentation.

## 2. Variables remain available after a cell finishes

### This is the single most important thing to understand about notebooks: **when you run a cell, any variable you create stays in memory.**

More precisely : your notebook is connected to a **Python session**, commonly called the
**kernel** (in Colab, the *runtime*). Every code cell you run talks to that same session, so
variables written by one cell can be read by any cell you run afterwards.

👉 Run the next two cells **in order** (top → bottom).

````python
# Cell A — create a variable
hand = "Each hand"
num_fingers = 5
````

````python
# Cell B — use the variables created in Cell A
print(f"{hand} has {num_fingers} fingers.")
````

````python
print(num_fingers)
````

Cell B worked because `hand` and `num_fingers` were already in the Python session,
put there by Cell A.

**Key idea:** the notebook has *one shared memory*. Every code cell reads from and writes to it.

---
## 3. Variable remain until the session restarts

That shared memory is not permanent. It is emptied whenever the Python session stops:

- you choose the dropdown a the right of **Run all** (Colab)
- Colab disconnects you after a period of inactivity (roughly 90 minutes)
- you close the tab and come back the next day

When that happens, **every variable disappears**, even though the code and the old outputs
are still visible on screen. This surprises everyone the first time: the notebook *looks*
full of results, but the session behind it is empty.

Your *file* is safe, it is saved in Google Drive. 

However, the *state* (variables, loaded data) are not available anymore until you run the notebook cells another time.

## 4. The execution number 

After you run a cell, a number appears in brackets to its bottom left:

```
[1]  ← the first cell you ran
[2]  ← the second cell you ran
...
[15] ← the fifteenth
```

This is **not** the cell's position in the notebook, it is the **order in which you ran it**.
A cell you have never run has an empty bracket `[ ]`, and a cell that is currently running shows `[*]`. 

Even if a cell has already been runned, its number will change as it is a global count (number of cell executed in the whole notebook). 

👉 Run the cell below **three times in a row** and watch the number change.

````python
print("Run me multiple times and watch the number on the left!")
````

The number goes up because each execution is a new step in the session's history.

If you see a cell with `[14]` above a cell and `[3]` below it, the top cell was run **more recently**
than the bottom one which also tells us that the notebook was not run in order. 

**This matters a LOT** as this can change the expected result of the notebook.

## 5. The most common notebook mistake: running cells out of order

Unlike a script (which always runs top to bottom), in a notebook you can click on any cell and
run it. This freedom is powerful, and it is also the number one source of confusion.

The three cells below are meant to be run A → B → C.

We will instead do the following *in this exact order*, and don't fix anything along the way:

1. Run **Cell C** first : Read the error message.
2. Run **Cell A**.
3. Run **Cell C** again : Does it work now?
4. Run **Cell B**.
5. Run **Cell C** one last time.

Then read the explanation below the cells.

````python
# --- Cell A ---
sample_rate = 30000  # Hz
````

````python
# --- Cell B ---
duration = 2  # seconds
n_samples = sample_rate * duration
````

````python
# --- Cell C ---
print(f"Number of samples: {n_samples}")
````

**What happened, step by step**

| Step | Result | Why |
|------|--------|-----|
| C alone | 🔴 `NameError: name 'n_samples' is not defined` | nothing had created `n_samples` yet running Cell A then B will create it.|
| A, then C | 🔴 still the same error | Cell A only creates `sample_rate`. `n_samples` is made by cell **B** |
| B, then C | ✅ `Number of samples: 60000` | Cell B could finally compute `n_samples`, because A had run |

Notice that **B itself only works after A has run**. Defining the `sample_rate` variable is required first. 

The Cells form a chain, and a notebook will happily let you skip a link in it even if this creates errors.

### About errors

Errors exists so Python interpreter can tell you what failed. Programmers see them every day. It is part of the process.

We should not fear errors. We can view them as a guidance to improve the code.

Read the **last line** first: it names the problem
(`NameError`), and the thing that caused it (`'n_samples'`). That is usually enough to fix it.

### Three habits that prevent almost all of this

- **Run cells top to bottom**, in order, the first time through.
- **Check the execution numbers** `[n]` when something looks odd.
- **When in doubt, restart and run all** (Section 9). We can see this as our last option (cold-restart)

## 6. Seeing results : `print()` vs variable as the last line

A notebook has a convenience: the **last expression** of a cell is displayed automatically,
without `print()`.

👉 Run the next two cells and compare their outputs.

````python
x = 42

x          # the last line of the cell → displayed automatically
````

````python
# Anything that is NOT the last line is computed, then silently thrown away
x          # ← you will NOT see this one
x + 10     # ← the last line, so this is the only value displayed
````

````python
# With print(), you see everything you ask for
print(x)
print(x + 10)
print("x is", x)
````

Good to know :

⚠️ `print()` **displays** a value, it does not **save** it. `print(x + 10)` shows `52`, but `x`
is still `42` afterwards. 

To keep a result, put it in a variable: `y = x + 10`.

## 7. Plot and figures

One of the best reasons to use notebooks: **plots appear directly below the code** that created
them. No separate windows, no saving files.

Don't worry about understanding every line below.

The main objective here is simply to see that a figure shows up underneath the cell.

````python
import numpy as np                  # library for numbers and arrays
import matplotlib.pyplot as plt     # library for plotting

frequency = 8                       # oscillations per second (Hz)  ← the number to change

time = np.linspace(0, 1, 1000)                                              # 1000 points over 1 second
signal = np.sin(2 * np.pi * frequency * time)  # wave

plt.figure(figsize=(8, 3))
plt.plot(time, signal, linewidth=0.5)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.title(f"Sinusoidal Signal ({frequency} Hz oscillation )")
plt.tight_layout()
plt.show()
````

👉 **Try changing something**

change the value of the variable to `frequency = 20` from `frequency = 8` then re-run the cell.

We can also display an Image pretty easily which can be a nice features in some specific cases. Without going into the details, IPython notebooks offer multiples rendering tools such as the one used here.

````python
# A notebook can also display images, not only text and plots.
from IPython.display import Image, display

cat_url = "https://upload.wikimedia.org/wikipedia/commons/3/3a/Cat03.jpg"

display(Image(url=cat_url, width=400))   # the image is fetched from the internet and shown below
````

## 8. Comments vs Markdown cells

Beginners often mix these two up. They are both "text", but they live in different places:

| Tool | Where it goes | What the reader sees |
|------|---------------|----------------------|
| `# comment` | inside a **code** cell | plain grey text, only visible in the code |
| **Markdown cell** | its own cell | formatted text: titles, **bold**, tables, links, images, and even LaTex formatted math equations.|

Use a `#` comment for a short note about *what a line does or why*.
Use a Markdown cell for section titles and real explanations.

## 9. Running everything: *Restart and run all*

So far you have run cells one at a time. Eventually you will want to run the whole notebook
in one go.

In Colab:

```
Click the ▼ dropdown beside “Run All” on the right, then select “Restart Session and Run All”
```

This does two things: it **empties the Python session** (all variables gone), then **runs every
cell from top to bottom**, in order.

**Why it matters:** it is the only way to prove your notebook actually works. If it runs cleanly
from a fresh session, then anyone can reproduce your results. 

If it
breaks, you have just found that something is problematic in your notebook.

> **Do this before submitting or sharing any analysis.**

## 10. How is a notebook different from a script?

There are three common ways to run Python. Each has its place.

| | **Notebook** (`.ipynb`) | **Script** (`.py` file) |
|--|------------------------|------------------------|
| **Runs how?** | Cell by cell, any order | Top to bottom, all at once |
| **Good for** | Exploring data, first drafts, teaching | Code that must run reliably |
| **Plots** | Appear inline | Open in a separate window |
| **Reproducibility** | ⚠️ Risky if cells are run out of order | ✅ Always the same order |
| **Sharing** | Ideal for testing and experimental code + text + plots in one file | Code only |

### In practice

- **Use a notebook** when you're exploring data, building figures, or writing an analysis you
  want to share. (More as a draft or scratch work)
- **Use a script** when you have a pipeline that must run reliably every time (for example,
  performing exhaustive dataset automatic processing).
- Many researchers **start in a notebook**, then move the stable parts into a `.py` script.

---
## 11. Strengths and weaknesses of notebooks

### ✅ Strengths

- **Instant feedback:** run a cell, see the result, adjust, repeat.
- **Documentation built in:** explanations and code in one document.
- **Great for figures:** plots live next to the code that creates them.
- **Easy to share:** send one `.ipynb` file and the reader sees everything.
- **Google Colab is free:** nothing to install, runs in your browser, free GPU / computing ressources for machine learning.

### ❌ Weaknesses

- **Hidden state:** a notebook can *appear* to work even though a required cell was never run. The variable is simply left over in memory from something you did earlier. *Restart and run all*
  is the cure.
- **Hard to debug long workflows:** finding a bug in a 50-cell analysis can be painful.
- **Not great for reusable code:** functions you want to reuse belong in `.py` files.
- **Version control is messy:** notebooks are stored as JSON, which is hard to review in Git.
- **Temptation to keep old cells:** stale code clutters the notebook and misleads readers.
- This might leads to a bloated notebook which miss the whole purpose of being simple and easy to use.

## Summary

| Concept | One-sentence takeaway |
|---------|----------------------|
| Cells | A notebook is a sequence of code cells and text (Markdown) cells. |
| Execution | `Shift+Enter` runs a cell; all cells share one Python session (the kernel). |
| Variables | Created in one cell, available to every cell you run afterwards. |
| Restarting | Restarting the session erases every variable — the outputs on screen stay, the memory doesn't. |
| Execution order | Cells can be run in any order: powerful **and** dangerous. |
| Restart and run all | Empties memory and runs everything top to bottom. The proof your notebook works. |
| Output | The last expression displays automatically; use `print()` to see several things. |
| Comments vs Markdown | `#` for short notes in code, Markdown cells for real explanations. |
| Notebooks vs scripts | Notebooks for exploring, scripts for pipelines. Both are useful. |

## Remember these three rules

1. **Run cells from top to bottom.**
2. **Variables stay in memory until the session restarts**. Then they are all gone.
3. **If something seems strange, restart and run all.**

Those three rules prevent most of the trouble beginners have with notebooks.
